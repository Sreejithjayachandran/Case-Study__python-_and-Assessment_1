```python
employees = [
    {"id": 101, "name": "Alice", "department": "Data", "salary": 65000, "rating": 4.5},
    {"id": 102, "name": "Bob", "department": "AI", "salary": 72000, "rating": 4.8},
    {"id": 103, "name": "Charlie", "department": "Web", "salary": 58000, "rating": 3.9},
    {"id": 104, "name": "David", "department": "Data", "salary": 61000, "rating": 4.2},
    {"id": 105, "name": "Eva", "department": "AI", "salary": 75000, "rating": 4.9},
    {"id": 106, "name": "Frank", "department": "Cloud", "salary": 68000, "rating": 4.1},
    {"id": 107, "name": "Grace", "department": "Web", "salary": 59000, "rating": 3.8},
    {"id": 108, "name": "Helen", "department": "Data", "salary": 70000, "rating": 4.6}
]

search_id = int(input("Enter Employee ID: "))

# 1. Search for the employee
found_employee = None
for emp in employees:
    if emp["id"] == search_id:
        found_employee = emp
        break.

if not found_employee:
    print(f"Employee Not Found")
else:
    # 3. Company calculations
    total_employees = len(employees)
    salaries = [emp["salary"] for emp in employees]
    avg_salary = sum(salaries) / total_employees
    highest_salary = max(salaries)
    lowest_salary = min(salaries)

    # Print  company data
    print("\n")
    print(f"Total Employees: {total_employees}")
    print("\n")
    print(f"Average Salary: {avg_salary}")
    print(f"Highest Salary: {highest_salary}")
    print(f"Lowest Salary: {lowest_salary}")

    # Eligible employees based on salary and rating
    print("\nEmployees with Salary > 65000 and Rating >= 4.5:")
    for emp in employees:
        if emp["salary"] > 65000 and emp["rating"] >= 4.5:
            print(f"{emp['name']}")

    # Employee ID–Name dictionary
    id_name_dict = {emp["id"]: emp["name"] for emp in employees}
    print(f"\nEmployee Dictionary\n{id_name_dict}")

    # Sorted employee names
    names = [emp["name"] for emp in employees]
    names.sort(reverse=True)
    print("\n")
    print("First Three Employee Names")
    for name in names[:3]:
        print(name)
    #Employee ID tuple details
    id_tuple = tuple(emp["id"] for emp in employees)
    print("\n")
    print("Employee ID Tuple:")
    print(f"{id_tuple}")

    #Max and Min
    print("\n")
    print(f"Maximum ID: {max(id_tuple)}")
    print(f"Minimum ID: {min(id_tuple)}")


    #Unique departments
    unique_departments = {emp["department"] for emp in employees}
    print("\nDepartments")
    print(unique_departments)
    print("\n")

    # 4. Employee details print at the very bottom
    print(f"Employee Found!")
    print(f"ID: {found_employee['id']}")
    print(f"Name: {found_employee['name']}")
    print(f"Department: {found_employee['department']}")
    print(f"Salary: {found_employee['salary']:,}")
    print(f"Rating: {found_employee['rating']}")
```
