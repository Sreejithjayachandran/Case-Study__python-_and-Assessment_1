```python
# Question 1
# Taking customer details from user
customer_name = input("Customer_Name: ")
units_consumed = int(input("Units_Consumed: "))
print("\n")

# Check  the units entered is a positive number
if units_consumed >= 1:
    # Case 1: 1 to 100 units
    if units_consumed <= 100:
        charge = units_consumed * 2

    # Case 2: 101 to 200 units
    elif units_consumed <= 200:
        charge = 100 * 2 + (units_consumed - 100) * 3

    # Case 3: 201 to 400 units
    elif units_consumed <= 400:
        charge = 100 * 2 + 100 * 3 + (units_consumed - 200) * 5

    # Case 4: Above 400 units
    else:
        charge = 100 * 2 + 100 * 3 + 200 * 5 + (units_consumed - 400) * 7

    # Calculate final charges
    service_charge = 50
    total_bill = charge + service_charge

    # Printing the final bill details
    print(f"Customer: {customer_name}")
    print(f"Units Consumed: {units_consumed}")
    print(f"Electricity Charge: {charge}")
    print(f"Service Charge: {service_charge}")
    print(f"Total Bill: {total_bill}")
else:
    # Check zero or negative unit inputs
    print("Invalid units")


```

```python
# Question 2
# Check the total number of students from the user
num_students = int(input())

# Lists to store user inputs
student_lines = []
for _ in range(num_students):
    student_lines.append(input())

#initilize values
passed_count = 0
failed_count = 0
highest_average = -1.0
top_performers = []

print("\n")


for line in student_lines:
    data = line.split()
    name = data[0]
    python = int(data[1])
    database = int(data[2])
    aptitude = int(data[3])

    # Calculate average
    average = (python + database + aptitude) / 3

    # Check: marks >= 40 and average >= 50
    if python >= 40 and database >= 40 and aptitude >= 40 and average >= 50:
        passed_count += 1

        # Assigning the grade
        if average >= 80:
            grade = "A"
        elif average >= 70:
            grade = "B"
        else:
            grade = "C"

        print(f"{name} - Average: {average:.2f} - Passed - Grade: {grade}")
    else:
        failed_count += 1
        print(f"{name} - Average: {average:.2f} - Failed")

    # find the top performers and handle equals
    if average > highest_average:
        highest_average = average
        top_performers = [name]
    elif average == highest_average:
        top_performers.append(name)

# Print final result
print(f"Passed: {passed_count}")
print(f"Failed: {failed_count}")
print(f"Top Performer: {', '.join(top_performers)}")
```

```python
# Question 3

class Product:
    def __init__(self, product_id, name, price, stock):
        self.product_id = product_id
        self.name = name
        self.price = float(price)
        self.stock = int(stock)


class ShoppingCart:

    def __init__(self, catalog):

        self.catalog = catalog

        self.items = {}

    def add_product(self, product_id, quantity):

        if product_id not in self.catalog:
            return

        if quantity <= 0:
            return

        product = self.catalog[product_id]
        current_qty = self.items.get(product_id, 0)


        if product.stock < (current_qty + quantity):
            print('Insufficient stock')
            return

        self.items[product_id] = current_qty + quantity

    def remove_product(self, product_id, quantity):

        if product_id not in self.items:
            print('Product not in cart')
            return

        if quantity >= self.items[product_id]:
            del self.items[product_id]
        else:
            self.items[product_id] -= quantity

    def calculate_total(self):

        total = 0.0
        for pid, qty in self.items.items():
            total += self.catalog[pid].price * qty
        return total

    def apply_discount(self, total_amount):

        if total_amount >= 20000:
            discount_pct = 15
        elif total_amount >= 10000:
            discount_pct = 10
        elif total_amount >= 5000:
            discount_pct = 5
        else:
            discount_pct = 0

        discount_amount = total_amount * (discount_pct / 100.0)
        final_amount = total_amount - discount_amount
        return discount_pct, discount_amount, final_amount

    def display_cart(self):

        for pid, qty in self.items.items():
            product = self.catalog[pid]
            subtotal = product.price * qty
            print(f'{product.name} - Quantity: {qty} - '
                  f'Subtotal: {int(subtotal)}')

        total_amount = self.calculate_total()
        disc_pct, disc_amt, final_amt = self.apply_discount(total_amount)

        print(f'Total Amount: {int(total_amount)}')
        print(f'Discount: {disc_pct}%')
        print(f'Discount Amount: {int(disc_amt)}')
        print(f'Final Amount: {int(final_amt)}')






```
